```python
import numpy as np
def load_ks_data(uu):
    X, Y = [], []
    for i in range(uu.shape[1] - 1):
        X.append(uu[:, i])    # u(t)
        Y.append(uu[:, i + 1])  # u(t+dt)
    return np.array(X), np.array(Y)
```

```python
X,Y = load_ks_data(uu)
```

```python
import torch.nn as nn
from collections import OrderedDict
```

```python
model = nn.Sequential(OrderedDict([
    ('linear1', nn.Linear(1024,512)),
    ('drop1', nn.Dropout(0.3)),
    ('relu1', nn.ReLU()),
    ('linear2', nn.Linear(512,256)),
    ('relu2', nn.ReLU()),
    ('drop2', nn.Dropout(0.2)),
    ('linear3', nn.Linear(256,128)),
    ('relu3', nn.ReLU()),
    ('output', nn.Linear(128,1024)),
]))
```
