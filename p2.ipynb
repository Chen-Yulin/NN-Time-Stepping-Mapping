```python
import scipy.io
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from tqdm import tqdm
import os
import pandas as pd

# Load and prepare data
mat = scipy.io.loadmat('/home/cyl/Datasets/DataMiningCourse/RD/reaction_diffusion_big.mat')
u = mat['u']
v = mat['v']
assert u.shape == v.shape
print("Data shape:", u.shape)
```

```python
def load_rd_data(u, v):
    """
    Load reaction-diffusion data with both u and v components as separate channels.

    Args:
        u: u component data of shape (spatial_dim, time)
        v: v component data of shape (spatial_dim, time)

    Returns:
        X: Input data with both u and v at time t, shape (time-1, spatial_dim*2)
        Y: Target data with both u and v at time t+1, shape (time-1, spatial_dim*2)
    """
    X, Y = [], []
    for i in range(u.shape[1] - 1):
        # Concatenate u and v components for input (time t)
        current_X = np.zeros(u.shape[0] * 2)
        for j in range(u.shape[0]):
            current_X[j] = u[j, i]              # u component at time t
            current_X[j + u.shape[0]] = v[j, i]  # v component at time t
        X.append(current_X)

        # Concatenate u and v components for target (time t+1)
        current_Y = np.zeros(u.shape[0] * 2)
        for j in range(u.shape[0]):
            current_Y[j] = u[j, i+1]              # u component at time t+1
            current_Y[j + u.shape[0]] = v[j, i+1]  # v component at time t+1
        Y.append(current_Y)

    return np.array(X), np.array(Y)
```
